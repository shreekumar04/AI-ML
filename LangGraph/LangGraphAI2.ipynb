Objective: 1. Use HumanMessage & AIMessage 
           2. Maintain a conversation history 
           3. Create a conversation loop

```python
import os
from typing import TypedDict, List, Union
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage, AIMessage
from langchain_openai import ChatOpenAI
from langgraph.graph import END, START, StateGraph
load_dotenv(override=True)
```

```python
class State(TypedDict):
    messages: List[Union[HumanMessage, AIMessage]] # Union function allows to keep both Human & AI messages
```

```python
llm = ChatOpenAI(model = "gpt-4o-mini")
```

```python
def process(state: State) -> State:
    response = llm.invoke(state["messages"])
    state["messages"].append(AIMessage(content = response.content))
    print(f"AI:{response.content}")
    return state
```

```python
graph = StateGraph(State)
graph.add_node("Node", process)
graph.add_edge(START, "Node")
graph.add_edge("Node", END)
agent = graph.compile()
```

```python
# Memory
conversation_history = []
user_input = input("Enter: ")
while user_input != "exit":
    conversation_history.append(HumanMessage(content = user_input))
    result = agent.invoke({"messages": conversation_history})
    conversation_history = result["messages"]
    user_input = input("Enter:")
```
